```python
import pandas as pd

# Archivo cargado en el entorno
archivo = 'TABLA DATOS ENCUESTA COMPLETA 2024.xlsx'

# Lectura de las 3 pestañas
df_integrantes = pd.read_excel(archivo, sheet_name='Integrantes del hogar')
df_hogar = pd.read_excel(archivo, sheet_name='Información del hogar')
df_codigos = pd.read_excel(archivo, sheet_name='Códigos')

# Confirmación de registros y columnas
print("Pestaña Integrantes:", df_integrantes.shape)
print("Pestaña Hogar:", df_hogar.shape)
print("Pestaña Códigos:", df_codigos.shape)
```

```python
import numpy as np

# 1. Reemplazo masivo de códigos especiales por nulos (NaN)
valores_no_respuesta = [997, 998, '997', '998', '997 - NS', '998 - NC', 'N/A', -221.555556]

df_integrantes_clean = df_integrantes.replace(valores_no_respuesta, np.nan)
df_hogar_clean = df_hogar.replace(valores_no_respuesta, np.nan)

# 2. Unificación relacional entre la tabla de integrantes y la de hogar usando la clave relacional
df_completo = pd.merge(
    df_integrantes_clean,
    df_hogar_clean,
    on='Clave',
    how='left',
    suffixes=('_integrante', '_hogar')
)

print("Dimensiones de la base unificada:", df_completo.shape)
print("Número total de variables consolidadas:", len(df_completo.columns))
```

```python
# 1. Resumen de percepción de calidad del agua para beber
reporte_agua = df_completo['PRESENCIA CALIDAD AGUA'].value_counts(dropna=False)
print("--- Percepción de Calidad de Agua ---")
print(reporte_agua)

# 2. Principales padecimientos y enfermedades reportadas por los integrantes
reporte_salud = df_completo['C1 Padecimientos'].value_counts().head(10)
print("\n--- Principales Padecimientos Reportados ---")
print(reporte_salud)

# 3. Distribución de la población encuestada por localidad y sexo
reporte_localidad = df_completo.groupby('Localidad_integrante')['C1 Sexo_integrante'].value_counts().unstack()
print("\n--- Distribución por Localidad y Sexo ---")
print(reporte_localidad)

# 4. Exportar la base de datos unificada limpia
df_completo.to_csv('base_encuesta_completa_unificada.csv', index=False)
print("\nBase unificada exportada como 'base_encuesta_completa_unificada.csv'")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Configuración de estilo gráfico
sns.set_theme(style="whitegrid")

# 1. Tabla cruzada: Padecimientos clave por localidad
padecimientos_clave = ['E. Hipertensión', 'B. Diabetes', 'C. Enfermedad renal']
ct_salud = pd.crosstab(
    df_completo['Localidad_integrante'],
    df_completo['C1 Padecimientos']
).reindex(columns=padecimientos_clave).fillna(0)

print("--- Prevalencia de Padecimientos por Localidad ---")
print(ct_salud)

# 2. Generación y exportación de gráfica de padecimientos
plt.figure(figsize=(12, 6))
ax = ct_salud.plot(kind='bar', figsize=(12, 6), width=0.8)
plt.title('Prevalencia de Padecimientos por Localidad', fontsize=14, fontweight='bold')
plt.xlabel('Localidad', fontsize=12)
plt.ylabel('Número de Casos Reportados', fontsize=12)
plt.xticks(rotation=35, ha='right')
plt.legend(title='Padecimiento')
plt.tight_layout()
plt.savefig('fig_padecimientos_localidad.png', dpi=300)
plt.show()

# 3. Tabla cruzada: Fuente de agua entubada por localidad
ct_agua = pd.crosstab(
    df_completo['Localidad_integrante'],
    df_completo['1 Agua entubada'],
    margins=True
)
print("\n--- Cobertura de Agua Entubada por Localidad ---")
print(ct_agua)

# 4. Perfil sociolaboral: Ocupación principal por localidad (Análisis socioagrícola/laboral)
ct_ocupacion = pd.crosstab(
    df_completo['Localidad_integrante'],
    df_completo['OCUPACIÓN - FINAL']
)
print("\n--- Distribución de Ocupaciones Principales por Localidad ---")
print(ct_ocupacion)
```

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Perfil de Exposición a Sustancias Tóxicas por Localidad
ct_toxicos = pd.crosstab(
    df_completo['Localidad_integrante'],
    df_completo['1 Tóxicos uso localidad'],
    normalize='index'
) * 100

print("--- % Uso/Presencia de Tóxicos en la Localidad ---")
print(ct_toxicos.round(1))

# 2. Prueba Estadística Chi-cuadrada de Independencia (Localidad vs. Padecimientos)
contingencia_salud = pd.crosstab(df_completo['Localidad_integrante'], df_completo['C1 Padecimientos'])
chi2, p_val, dof, _ = stats.chi2_contingency(contingencia_salud)

print("\n--- Resultado de Prueba Chi-cuadrada (Significancia Estadística) ---")
print(f"Estadístico Chi2: {chi2:.2f}")
print(f"p-value: {p_val:.4e} {'(Estadísticamente significativo p < 0.05)' if p_val < 0.05 else ''}")

# 3. Gráfica de Exposición Ambiental por Localidad
plt.figure(figsize=(10, 5))
sns.barplot(data=df_completo, x='Localidad_integrante', y=df_completo['1 Tóxicos uso localidad'] == 'SÍ', ci=None, palette='Blues_d')
plt.title('Proporción de Hogares que Reportan Uso/Presencia de Tóxicos en la Localidad', fontsize=13, fontweight='bold')
plt.ylabel('Proporción (0 a 1)', fontsize=11)
plt.xlabel('Localidad', fontsize=11)
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.savefig('fig_exposicion_toxicos_localidad.png', dpi=300)
plt.show()

# 4. Exportación del Libro Excel de Reporte Consolidado (Múltiples Pestañas)
with pd.ExcelWriter('resumen_indicadores_tlaxcala_puebla.xlsx', engine='openpyxl') as writer:
    pd.crosstab(df_completo['Localidad_integrante'], df_completo['C1 Sexo_integrante']).to_excel(writer, sheet_name='Demografia')
    contingencia_salud.to_excel(writer, sheet_name='Padecimientos')
    pd.crosstab(df_completo['Localidad_integrante'], df_completo['1 Tóxicos uso localidad']).to_excel(writer, sheet_name='Exposicion_Toxicos')
    pd.crosstab(df_completo['Localidad_integrante'], df_completo['1 Agua entubada']).to_excel(writer, sheet_name='Agua_Entubada')

print("\nLibro consolidado generado con éxito: 'resumen_indicadores_tlaxcala_puebla.xlsx'")
```
